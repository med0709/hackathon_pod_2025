## Carregamento de dados do Bucket

```python
# autenticação via Resource Principal do Notebook
signer = oci.auth.signers.get_resource_principals_signer()

# cliente do Object Storage usando o signer do ambiente
client = oci.object_storage.ObjectStorageClient({}, signer=signer)

# obtém o namespace automaticamente
namespace = client.get_namespace().data

print("Namespace:", namespace)
```

```python
# bucket e objeto permanecem iguais
bucket_name = "data"
object_name = "raw/book_variaveis_04_v2.parquet"

# define diretório absoluto (recomendado)
target_dir = "/home/datascience/ciencia_dados/data/raw"
target_file = os.path.join(target_dir, "book_variaveis.parquet")

# cria a pasta se não existir
os.makedirs(target_dir, exist_ok=True)

# baixa do bucket
obj = client.get_object(namespace, bucket_name, object_name)

# salva no caminho correto
with open(target_file, "wb") as f:
    f.write(obj.data.content)

print("Arquivo salvo em:", target_file)
```
